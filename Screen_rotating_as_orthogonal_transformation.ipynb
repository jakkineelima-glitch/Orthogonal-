```python

# ============================================================
# CELL 4 — MATHEMATICAL MODEL
# ============================================================

# Standard 2-D rotation matrix:
#
# Q(θ) = [ cosθ  -sinθ ]
#        [ sinθ   cosθ ]
#
# This matrix rotates a point by θ degrees.

theta = sp.symbols('theta', real=True)

Q_general = sp.Matrix([
    [sp.cos(theta), -sp.sin(theta)],
    [sp.sin(theta),  sp.cos(theta)]
])

print("GENERAL ROTATION MATRIX")
print("-" * 35)

sp.pprint(Q_general)

print("\nMathematical transformation:")
print("x' = Qx")

print("\nFor a 90° rotation:")
print("θ = 90°")
print("cos(90°) = 0")
print("sin(90°) = 1")
```

```python
# ============================================================
# CELL 5 — ROTATION MATRIX
# ============================================================

theta_value = np.radians(ROTATION_ANGLE)

Q = np.array([
    [np.cos(theta_value), -np.sin(theta_value)],
    [np.sin(theta_value),  np.cos(theta_value)]
])

print("ROTATION MATRIX")
print("-" * 35)

print("Q =")
print(Q)

print("\nFor 90°:")
if np.isclose(ROTATION_ANGLE % 360, 90):
    print("Q = [[0, -1],")
    print("     [1,  0]]")
```

```python
# ============================================================
# CELL 6 — ORTHOGONALITY AND DETERMINANT
# ============================================================

QT = Q.T
QTQ = QT @ Q
I = np.eye(2)

det_Q = np.linalg.det(Q)

print("ORTHOGONALITY VERIFICATION")
print("-" * 40)

print("Qᵀ =")
print(QT)

print("\nQᵀQ =")
print(QTQ)

print("\nIdentity matrix I =")
print(I)

orthogonal = np.allclose(QTQ, I)

print("\nQᵀQ = I ?", orthogonal)

print("\nDeterminant:")
print(f"det(Q) = {det_Q:.6f}")

print("\nConclusion:")

if orthogonal:
    print("✓ Q is an orthogonal matrix.")

if np.isclose(det_Q, 1):
    print("✓ det(Q) = +1, so this is a pure rotation.")

elif np.isclose(det_Q, -1):
    print("✓ det(Q) = -1, indicating an orthogonal transformation")
    print("  involving reflection.")
```

```python
# ============================================================
# REUSABLE ROTATION FUNCTIONS
# ============================================================

def rotation_matrix(angle_degrees):
    """
    Return the 2-D rotation matrix for the given angle.
    """
    angle = np.radians(angle_degrees)

    return np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle),  np.cos(angle)]
    ])


def rotate_point(point, angle_degrees):
    """
    Rotate a 2-D point by angle_degrees.
    """
    Q = rotation_matrix(angle_degrees)
    point = np.asarray(point, dtype=float)

    return Q @ point


# Example
test_angle = 90
test_point = np.array([300.0, 500.0])

result = rotate_point(test_point, test_angle)

print("Reusable Function Test")
print("-" * 30)
print("Original:", test_point)
print("Angle:", test_angle, "degrees")
print("Rotated :", result)
```

```python
# ============================================================
# CELL 7 — COORDINATE TRANSFORMATION
# ============================================================

point = np.array([
    X_COORDINATE,
    Y_COORDINATE
], dtype=float)

# Mathematical equation:
#
# x' = Qx
#
# Python:
# rotated_point = Q @ point

rotated_point = Q @ point

print("COORDINATE TRANSFORMATION")
print("-" * 40)

print("Original coordinate:")
print(point)

print("\nTransformation:")
print("x' = Qx")

print("\nRotated coordinate:")
print(rotated_point)

print(
    f"\n({point[0]:.2f}, {point[1]:.2f}) "
    f"→ "
    f"({rotated_point[0]:.2f}, {rotated_point[1]:.2f})"
)
```

```python
# ============================================================
# CELL 8 — DISTANCE + QUADRATIC FORM INVARIANCE
# ============================================================

# Distance from origin:
# ||x|| = sqrt(xᵀx)

original_distance = np.linalg.norm(point)
rotated_distance = np.linalg.norm(rotated_point)

# Quadratic form:
# xᵀx

quadratic_before = point.T @ point
quadratic_after = rotated_point.T @ rotated_point

print("DISTANCE PRESERVATION")
print("-" * 40)

print(f"Original distance : {original_distance:.6f}")
print(f"Rotated distance  : {rotated_distance:.6f}")

print("\nDistance preserved:",
      np.isclose(original_distance, rotated_distance))

print("\nQUADRATIC FORM INVARIANCE")
print("-" * 40)

print(f"xᵀx       = {quadratic_before:.6f}")
print(f"(Qx)ᵀ(Qx) = {quadratic_after:.6f}")

print("\nMathematical proof:")
print("(Qx)ᵀ(Qx)")
print("= xᵀQᵀQx")
print("= xᵀIx")
print("= xᵀx")

print("\nQuadratic form preserved:",
      np.isclose(quadratic_before, quadratic_after))
```

```python
# ============================================================
# CELL 9 — TWO SUCCESSIVE ROTATIONS
# ============================================================

Q_squared = Q @ Q

print("TWO SUCCESSIVE ROTATIONS")
print("-" * 40)

print("Q² =")
print(Q_squared)

two_rotation_point = Q_squared @ point

print("\nOriginal point:")
print(point)

print("\nAfter two rotations:")
print(two_rotation_point)

print("\nFor two 90° rotations:")
print("Q² =")
print("[[-1,  0],")
print(" [ 0, -1]]")

print("\nThis is the standard 180° rotation matrix.")
```

```python
# ============================================================
# CELL 11 — AUTOMATIC TESTING
# ============================================================

def rotation_matrix(angle_degrees):
    """Create a standard 2-D rotation matrix."""

    angle = np.radians(angle_degrees)

    return np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle),  np.cos(angle)]
    ])


def rotate_point(point, angle_degrees):
    """Rotate a 2-D point."""

    matrix = rotation_matrix(angle_degrees)

    return matrix @ np.asarray(point, dtype=float)


print("AUTOMATIC BENCHMARK TEST")
print("=" * 65)

all_passed = True

for i, item in enumerate(benchmark_data, start=1):

    p = np.array(item["point"], dtype=float)
    angle = item["angle"]

    matrix = rotation_matrix(angle)
    result = matrix @ p

    # Orthogonality test
    orthogonal_test = np.allclose(
        matrix.T @ matrix,
        np.eye(2)
    )

    # Distance preservation
    distance_test = np.isclose(
        np.linalg.norm(p),
        np.linalg.norm(result)
    )

    passed = orthogonal_test and distance_test

    if not passed:
        all_passed = False

    print(
        f"Test {i}: "
        f"{'PASS ✓' if passed else 'FAIL ✗'}"
    )

    print(f"  Point: {p}")
    print(f"  Angle: {angle}°")
    print(f"  Result: {np.round(result, 4)}")
    print(f"  Distance preserved: {distance_test}")
    print()

print("=" * 65)

if all_passed:
    print("ALL BENCHMARK TESTS PASSED ✓")
else:
    print("SOME TESTS FAILED ✗")
```

```python
# ============================================================
# CELL 14 — ROTATION ANIMATION
# ============================================================

from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Use a smaller normalized screen for clear animation
animation_screen = np.array([
    [-2, -3.5],
    [ 2, -3.5],
    [ 2,  3.5],
    [-2,  3.5],
    [-2, -3.5]
])

fig, ax = plt.subplots(figsize=(7, 7))

line, = ax.plot([], [], linewidth=3)

angle_text = ax.text(
    0.02,
    0.95,
    "",
    transform=ax.transAxes,
    fontsize=14
)

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_aspect("equal")

ax.axhline(0, linewidth=0.7)
ax.axvline(0, linewidth=0.7)

ax.grid(True)

ax.set_title(
    "Live Phone Screen Rotation"
)


def update(frame):

    angle = frame

    matrix = rotation_matrix(angle)

    rotated = animation_screen @ matrix.T

    line.set_data(
        rotated[:, 0],
        rotated[:, 1]
    )

    angle_text.set_text(
        f"Rotation Angle: {angle}°"
    )

    return line, angle_text


frames = list(range(0, 361, 5))

animation = FuncAnimation(
    fig,
    update,
    frames=frames,
    interval=40,
    blit=True
)

plt.close(fig)

HTML(animation.to_jshtml())
```

```python
# ============================================================
# CELL 16 — ORTHOGONAL VS NON-ORTHOGONAL
# ============================================================

# Orthogonal transformation: rotation
Q_rotation = rotation_matrix(90)

# Non-orthogonal transformation: simple scaling
S_scaling = np.array([
    [2, 0],
    [0, 1]
])

original = np.array([3.0, 4.0])

rotated = Q_rotation @ original
scaled = S_scaling @ original

print("ORTHOGONAL VS NON-ORTHOGONAL")
print("=" * 55)

print("Original vector:")
print(original)

print("\nOrthogonal rotation:")
print(rotated)

print("Original length:",
      np.linalg.norm(original))

print("Rotated length:",
      np.linalg.norm(rotated))

print("\nNon-orthogonal scaling:")
print(scaled)

print("Scaled length:",
      np.linalg.norm(scaled))

print("\nConclusion:")
print("✓ Rotation preserves distance.")
print("✗ Scaling changes distance.")
print("\nTherefore, rotation is an orthogonal transformation,")
print("while this scaling transformation is not.")
```
